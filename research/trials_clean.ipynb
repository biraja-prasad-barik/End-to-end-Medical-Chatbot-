# Medical Chatbot Research

This notebook contains the research and development for the medical chatbot using RAG (Retrieval Augmented Generation).

```python
# Import required libraries
import os
from dotenv import load_dotenv

# Load environment variables
load_dotenv()

# Check if API keys are loaded
if os.environ.get('PINECONE_API_KEY'):
    print("✅ Pinecone API key loaded")
else:
    print("❌ Pinecone API key not found")

if os.environ.get('HUGGINGFACEHUB_API_TOKEN'):
    print("✅ HuggingFace API token loaded")
else:
    print("❌ HuggingFace API token not found")
```

```python
# Load PDF documents
from src.helper import load_pdf_file, text_split, download_hugging_face_embeddings

# Extract data from PDFs
extracted_data = load_pdf_file(data="Data")
print(f"Loaded {len(extracted_data)} documents")

# Split into chunks
text_chunks = text_split(extracted_data)
print(f"Created {len(text_chunks)} text chunks")
```

```python
# Load embeddings
embeddings = download_hugging_face_embeddings()
print("Embeddings loaded successfully")
```

```python
# Create vector store
from langchain_pinecone import PineconeVectorStore

index_name = "medicalbot"

# Create or load existing index
docsearch = PineconeVectorStore.from_existing_index(
    index_name=index_name,
    embedding=embeddings
)

print("Vector store connected successfully")
```

```python
# Test retrieval
retriever = docsearch.as_retriever(search_type="similarity", search_kwargs={"k": 3})
test_docs = retriever.invoke("What is diabetes?")

print(f"Retrieved {len(test_docs)} documents for test query")
if test_docs:
    print(f"Sample content: {test_docs[0].page_content[:200]}...")
```
